### Question 1 - The WorldLight Company (5 points)
The WorldLight Company produces two light fixtures (products 1 and 2) that require
both metal frame parts and electrical components. Management wants to determine
how many units of each product to produce so as to maximize profit. For each unit of
product 1, 1 unit of frame parts and 2 units of electrical components are required. For
each unit of product 2, 3 units of frame parts and 2 units of electrical components are
required. The company has 200 units of frame parts and 300 units of electrical
components. Each unit of product 1 gives a profit of €1, and each unit of product 2,
up to 60 units, gives a profit of €2. Any excess over 60 units of product 2 brings no

Your model should include:
two decision variables:
x1 for the number of units of product 1 produced
x2 for the number of units of product 2 produced
the constraints needed to represent the production limits in the problem
this must include a constraint named frame_parts, ensuring that the
total number of frame parts used does not exceed the number available
this must include a constraint named electrical_parts, ensuring that
the total number of electrical components used does not exceed the
number available
one objective named profit, representing the total profit to be maximized

### Question 2 - Weenies and Buns (5 points)
Weenies and Buns is a food processing plant which manufactures buns and
frankfurters for hot dogs. They grind their own flour for the buns at a maximum rate
of 200 pounds per week. Each bun requires 0.1 pound of flour. They currently have a
contract with Pigland, Inc., which specifies that a delivery of 800 pounds of pork
product is delivered every Monday. Each frankfurter requires 0.25 pound of pork
product. All the other ingredients in the buns and frankfurters are in plentiful supply.
Finally, the labor force at Weenies and Buns consists of 5 employees working full time
(40 hours per week each). Each frankfurter requires 3 minutes of labor, and each bun
requires 2 minutes of labor. Each frankfurter yields a profit of €0.88, and each bun
yields a profit of €0.33. Weenies and Buns would like to know how many frankfurters
and how many buns they should produce each week so as to achieve the highest
possible profit.
Your model should include:
two decision variables:
x1 for the number of frankfurters produced
x2 for the number of buns produced
the constraints needed to represent the production limits in the problem
this must include a constraint named working_hours, ensuring that the
total required working time does not exceed the total working time
available
one objective named profit, representing the total profit to be maximized

# Question 1 - The WorldLight Company (5 points)

```python
# DO NOT CHANGE IMPORTS (Used AI to get pyomo to import)
import sys
!{sys.executable} -m pip install pyomo highspy
import pyomo.environ as pyo
```

```python
#Create a model for the worldlight problem
def model_worldlight():
  model = pyo.ConcreteModel(name='worldlight')
  model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
  model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

  @model.Objective(sense=pyo.maximize)
  def profit(model):
    return model.x1 + (2 * model.x2)

  @model.Constraint()
  def demandlimit(model):
    return model.x2 <= 60

  @model.Constraint()
  def frame_parts(model):
    return model.x1 + (3 * model.x2) <= 200

  @model.Constraint()
  def electrical_parts(model):
    return (2 * model.x1) + (2 * model.x2) <= 300
  return model

#Solve the model and return the results
model = model_worldlight()

solver = pyo.SolverFactory("highs")
results = solver.solve(model)

print("Solver status:", results.solver.termination_condition)
print("Optimal product 1 :", pyo.value(model.x1))
print("Optimal product 2:", pyo.value(model.x2))
print("Maximum profit:", pyo.value(model.profit))
```

# Question 2 - Weenies and Buns (5 points)

```python
def model_wb():
  model = pyo.ConcreteModel(name='wb')
  model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
  model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

  @model.Objective(sense=pyo.maximize)
  def profit(model):
    return (0.88 * model.x1) + (0.33 * model.x2)

  @model.Constraint()
  def porklimit(model):
    return model.x1 * 0.25 <= 800

  @model.Constraint()
  def flourlimit(model):
    return 0.1 * model.x2 <= 200

  @model.Constraint()
  def working_hours(model):
    return (3 * model.x1) + (2 * model.x2) <= (5 * (60 * 40))
  return model

#Solve the model and return the results
model = model_wb()

solver = pyo.SolverFactory("highs")
results = solver.solve(model)

print("Solver status:", results.solver.termination_condition)
print("Optimal frankfurters:", pyo.value(model.x1))
print("Optimal buns:", pyo.value(model.x2))
print("Maximum profit:", pyo.value(model.profit))
```
