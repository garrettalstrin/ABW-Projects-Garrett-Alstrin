```python
#Create a model for the worldlight problem
def model_worldlight():
  model = pyo.ConcreteModel(name='worldlight')
  model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
  model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

  @model.Objective(sense=pyo.maximize)
  def profit(model):
    return model.x1 + (2 * model.x2)

  @model.Constraint()
  def demandlimit(model):
    return model.x2 <= 60

  @model.Constraint()
  def frame_parts(model):
    return model.x1 + (3 * model.x2) <= 200

  @model.Constraint()
  def electrical_parts(model):
    return (2 * model.x1) + (2 * model.x2) <= 300
  return model

#Solve the model and return the results
model = model_worldlight()

solver = pyo.SolverFactory("highs")
results = solver.solve(model)

print("Solver status:", results.solver.termination_condition)
print("Optimal x1:", pyo.value(model.x1))
print("Optimal x2:", pyo.value(model.x2))
print("Maximum profit:", pyo.value(model.profit))
```

# Question 2 - Weenies and Buns (5 points)

```python
def model_wb():
  model = pyo.ConcreteModel(name='wb')
  model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
  model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

  @model.Objective(sense=pyo.maximize)
  def profit(model):
    return (0.88 * model.x1) + (0.33 * model.x2)

  @model.Constraint()
  def porklimit(model):
    return model.x1 * 0.25 <= 800

  @model.Constraint()
  def flourlimit(model):
    return 0.1 * model.x2 <= 200

  @model.Constraint()
  def working_hours(model):
    return (3 * model.x1) + (2 * model.x2) <= (5 * (60 * 40))
  return model

#Solve the model and return the results
model = model_wb()

solver = pyo.SolverFactory("highs")
results = solver.solve(model)

print("Solver status:", results.solver.termination_condition)
print("Optimal x1:", pyo.value(model.x1))
print("Optimal x2:", pyo.value(model.x2))
print("Maximum profit:", pyo.value(model.profit))
```
