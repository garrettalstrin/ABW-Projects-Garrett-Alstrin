# Project Starting Point
# Optimizing Access to Health Centers in Timor-Leste
Timor-Leste, an island nation in Southeast Asia, faces significant infrastructure challenges, particularly
in healthcare. In this assignment, we will utilize analytics to identify optimal health center locations
in Ermera, a district in Timor-Leste.

To determine the optimal health center locations in Ermera, we will apply Mixed Integer Linear
Optimization. Specifically, we will model this as a Facility Location Problem, where the objective is to
maximize the number of households that can access a health center within 3 kilometers.
For simplicity, we will assume each household consists of one person


```python
import matplotlib.pyplot as plt


def ShowFacilityLocation(xHomes, yHomes, xFacilities, yFacilities, X=None, Z=None):
    """Plot candidate sites and households, optionally highlighting selected IDs."""
    home_points = pd.DataFrame({'lon': xHomes, 'lat': yHomes})
    facility_points = pd.DataFrame({'lon': xFacilities, 'lat': yFacilities})
    selected_sites = [] if X is None else list(X)
    covered_homes = [] if Z is None else list(Z)

    _fig, ax = plt.subplots(figsize=(8, 5))
    ax.scatter(home_points['lon'], home_points['lat'], s=8, c='lightgray',
               label='Households')
    ax.scatter(facility_points['lon'], facility_points['lat'], s=35,
               marker='s', facecolors='none', edgecolors='tab:blue',
               label='Candidate sites')
    if covered_homes:
        covered = home_points.loc[covered_homes]
        ax.scatter(covered['lon'], covered['lat'], s=12, c='tab:green',
                   label='Covered households')
    if selected_sites:
        opened = facility_points.loc[selected_sites]
        ax.scatter(opened['lon'], opened['lat'], s=55, marker='s',
                   c='gold', edgecolors='black', label='Opened sites')
    ax.set(xlabel='Longitude', ylabel='Latitude', title='Timor-Leste facility-location data')
    ax.legend()
    plt.show()
```

```python
ShowFacilityLocation(homes['lon'], homes['lat'],
                     locations['lon'], locations['lat'])
```

```python
import time

import pyomo.environ as pyo
from teaching_utils import make_solver

# 1. Define parameters and sets
new_facility_limit = 5          # Number of facilities to open
coverage_threshold = 3     # Service distance threshold in km

I = list(homes.index)       # Households
J = list(locations.index)   # Potential facility sites

# Precompute coverage: dij is 1 if facility j can cover household i
dij_less_then_coverage_threshold = {}
for i in I:
    for j in J:
        dij_less_then_coverage_threshold[i, j] = 1 if distances.at[i, j] <= coverage_threshold else 0

# Build the Pyomo Concrete Model
model = pyo.ConcreteModel(name="Maximal_Covering")

# Decision variables
model.xj = pyo.Var(J, domain=pyo.Binary)  # 1 if facility j is opened, 0 otherwise
model.zi = pyo.Var(I, domain=pyo.Binary)  # 1 if household i is covered, 0 otherwise

# Objective using decorator
@model.Objective(sense=pyo.maximize)
def obj(model):
    return sum(model.zi[i] for i in I)

# Constraint: Maximum number of facilities to open (new_facility_limit >= sigma(j:J) Xj)
@model.Constraint()
def facility_limit(model):
    return sum(model.xj[j] for j in J) <= new_facility_limit

# Constraint: Household i can only be covered if at least one serving facility is opened (Zi <= Sigma(j:dij<=3)Xj)
@model.Constraint(I)
def coverage_con(model, i):
    return model.zi[i] <= sum(dij_less_then_coverage_threshold[i, j] * model.xj[j] for j in J)

# Solve the model using the appsi_highs solver and wall time
solver = make_solver('appsi_highs')
t0 = time.perf_counter()
results = solver.solve(model, tee=True)
t1 = time.perf_counter()
print(f"Wall time: {t1 - t0:.2f} s")
print(f"Status: {results.solver.termination_condition}")

# Extract results
opened_sites = [j for j in J if pyo.value(model.xj[j]) > 0.5]
covered_homes = [i for i in I if pyo.value(model.zi[i]) > 0.5]

coverage_ratio = len(covered_homes) / len(I)
print(f"Selected facility sites: {opened_sites}")
print(f"Total households covered: {len(covered_homes)} out of {len(I)} ({coverage_ratio:.2%})")

# Visualize the solution
ShowFacilityLocation(homes['lon'], homes['lat'],
                     locations['lon'], locations['lat'],
                     X=opened_sites, Z=covered_homes)
```

```python
import time

import pyomo.environ as pyo
from teaching_utils import make_solver

I = list(homes.index)  # Households
J = list(locations.index)

model = pyo.ConcreteModel()

# --- Sets ---
model.I = pyo.Set(initialize=I)
model.J = pyo.Set(initialize=J)

# --- Parameters ---
# Vi: household coverage weight; each household counts once
model.Vi = pyo.Param(model.I, within=pyo.NonNegativeReals, default=1.0)

# dij: the travel distance from household i to health centre j
model.dij = pyo.Param(
    model.I,
    model.J,
    initialize=lambda model, i, j: float(distances.at[i, j]),
    within=pyo.NonNegativeReals,
)

# S: the max travel distance threshold
model.S = pyo.Param(within=pyo.NonNegativeReals, default=3.0)

# p: the max number of health centres that can be opened
model.p = pyo.Param(within=pyo.NonNegativeIntegers, default=5)


# --- Derived Sets ---
# Pairs: (i, j) pairs within the distance threshold S
def pairs_rule(model):
    return ((i, j) for i in model.I for j in model.J if model.dij[i, j] <= model.S)


model.Pairs = pyo.Set(dimen=2, initialize=pairs_rule)

# --- Decision Variables ---
# y[i, j] = 1 if household i is assigned to facility j, 0 otherwise (only defined for valid pairs)
model.y = pyo.Var(model.I, model.J, within=pyo.Binary)

# x[j] = 1 if facility j is opened, 0 otherwise
model.x = pyo.Var(model.J, within=pyo.Binary)


# --- Objective Function ---
def objective_rule(model):
    return sum(model.Vi[i] * model.y[i, j] for (i, j) in model.Pairs)


model.obj = pyo.Objective(rule=objective_rule, sense=pyo.maximize)


# --- Constraints ---
# 1. Each household can be assigned to AT MOST one health centre
def demand_rule(model, i):
    valid_centers = [j for j in model.J if (i, j) in model.Pairs]
    if not valid_centers:
        return pyo.Constraint.Skip
    return sum(model.y[i, j] for j in valid_centers) <= 1


model.demand = pyo.Constraint(model.I, rule=demand_rule)


# 2. A household can only be assigned to an OPEN health centre
def open_facility_rule(model, i, j):
    return model.y[i, j] <= model.x[j]


model.open_facility_tight = pyo.Constraint(model.Pairs, rule=open_facility_rule)


# 3. Maximum number of health centres that can be opened is p
def max_facilities_rule(model):
    return sum(model.x[j] for j in model.J) <= model.p


model.max_facilities = pyo.Constraint(rule=max_facilities_rule)

# Solve the model using the appsi_highs solver and wall time
solver = make_solver("appsi_highs")
t0 = time.perf_counter()
results = solver.solve(model, tee=True)
t1 = time.perf_counter()
print(f"Wall time: {t1 - t0:.2f} s")
print(f"Status: {results.solver.termination_condition}")

# Extract results
opened_sites = [j for j in J if pyo.value(model.x[j]) > 0.5]
covered_homes = [
    i
    for i in I
    if any(pyo.value(model.y[i, j]) > 0.5 for j in J if (i, j) in model.Pairs)
]

coverage_ratio = len(covered_homes) / len(I)
print(f"Selected facility sites: {opened_sites}")
print(
    f"Total households covered: {len(covered_homes)} out of {len(I)} ({coverage_ratio:.2%})"
)

# Visualize the solution
ShowFacilityLocation(
    homes["lon"],
    homes["lat"],
    locations["lon"],
    locations["lat"],
    X=opened_sites,
    Z=covered_homes,
)
```
